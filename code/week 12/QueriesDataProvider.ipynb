# Queries Data Provider
This file is used to parse the results for the additional queries of each week.<br>
This includes both inputData and outputs results for each week.

Including used libraries

```python
import numpy as np
import os
PREVIOUS_WEEK = 11
```

Custom functions definitions

```python
def parseInput(input):
    out = []
    line = []
    current = []
    buf = ''
    for x in input:
        if x == ',':
            if len(buf) > 0:
                current.append(float(buf))
                buf = ''
        else:
            if x == ']':
                if len(buf) > 0:
                    current.append(float(buf))
                    buf = ''
                    line.append(current)
                    current = []
                else:
                    out.append(line)
                    line = []
            else:
                if x != '\n' and x != ' ' and  x != '[':
                    buf += x
    return out

```

```python
def parseOutput(output):
#    print(output)
    out = []
    line = []
    buf = ''
    for x in output:
        if x == ',':
            if len(buf) > 0:
                line.append(np.float64(buf))
                buf = ''
        else:
            if x == ']':
                if len(buf) > 0:
                    line.append(np.float64(buf))
                    buf = ''
                    out.append(line)
                    line = []
                else:
                    out.append(line)
                    line = []
            else:
                if x != '\n' and x != ' ' and  x != '[':
                    buf += x
    return out
```

Data parsing

```python
inputFileName = './../../data/queries_data/week'+str(PREVIOUS_WEEK)+'/inputs.txt'
inputFile = open(inputFileName)
inputData = inputFile.read()
inputFile.close()

inputData = inputData.replace('array(', '')
inputData = inputData.replace(')', '')

inputs = []
for i in range(8):
    row = []
    inputs.append(row)

parsed = parseInput(inputData)
#print('result: ', parsed)
for line in parsed:
    for i,func in enumerate(line):
        inputs[i].append(func)
#print(inputs)

newpath = r'./parsed'
if not os.path.exists(newpath):
    os.makedirs(newpath)

for i,values in enumerate(inputs):
    print('function ',i + 1,' inputs: ', values)
    data = np.array(values)
    np.save(f'./parsed/function{i + 1}inputs.npy', data)

```

```python
outputFileName = './../../data/queries_data/week'+str(PREVIOUS_WEEK)+'/outputs.txt'
outputFile = open(outputFileName)
outputData = outputFile.read()
outputFile.close()

outputData = outputData.replace('np.float64(', '')
outputData = outputData.replace(')', '')

#print(outputData)

outputs = []
for i in range(8):
    row = []
    outputs.append(row)

parsed = parseOutput(outputData)
#print('result: ', parsed)
for line in parsed:
    for i,func in enumerate(line):
        outputs[i].append(func)
#print(outputs)

newpath = r'./parsed'
if not os.path.exists(newpath):
    os.makedirs(newpath)

for i,values in enumerate(outputs):
    print('function ',i + 1,' outputs: ', values)
    data = np.array(values)
    np.save(f'./parsed/function{i + 1}outputs.npy', data)

```
